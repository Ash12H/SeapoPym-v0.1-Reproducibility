# Twin experiments — run all 7 GA optimisations

Orchestrate `scripts/run_optimization.py` for the six single-station
experiments and the joint MERGED experiment (Manuscript Section 2.4.4).

The `mode` (test / production) is read from `parameters.yaml` unless the
`MODE` variable below is set explicitly.

Output: `data/ga_logbook_{STATION}_{MODE}.parquet` for each experiment.
Runtime: test mode ~1-2 min total; production mode ~7-14 h on a 30-core cluster.

```python
import subprocess
import sys
from pathlib import Path

import yaml


def _project_root(marker: str = "pyproject.toml") -> Path:
    for parent in [Path.cwd(), *Path.cwd().parents]:
        if (parent / marker).exists():
            return parent
    raise FileNotFoundError(f"Project root marker {marker!r} not found.")


PROJECT_ROOT = _project_root()
SCRIPT = PROJECT_ROOT / "scripts" / "run_optimization.py"

MODE = None  # set to "test" or "production" to override parameters.yaml
with open(PROJECT_ROOT / "parameters.yaml") as f:
    GA = yaml.safe_load(f)["genetic_algorithm"]
mode = MODE or GA["mode"]
experiments = GA["experiments"]
print(f"Mode: {mode} | experiments: {experiments}")
```

```python
for station in experiments:
    print(f"\n========== {station} (mode={mode}) ==========")
    cmd = [sys.executable, str(SCRIPT), "--station", station, "--mode", mode]
    result = subprocess.run(cmd, cwd=str(PROJECT_ROOT))
    if result.returncode != 0:
        raise RuntimeError(f"Optimization for {station} failed (exit code {result.returncode})")
print("\nAll experiments completed.")
```
